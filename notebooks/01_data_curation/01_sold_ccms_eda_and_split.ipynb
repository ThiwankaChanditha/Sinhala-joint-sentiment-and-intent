```python
import pandas as pd

df = pd.read_csv("/content/drive/MyDrive/SOLD dataset/ccms-sentence-level-annotation.csv")
```

```python
df = df[['Sentence', 'Sentiment']]
df.head()
```

```python
df['Sentence'].count()
```

```python
import re
import unicodedata
import pandas as pd

def clean_text(text):
    if not isinstance(text, str):
        return text

    # Handle Unicode: Convert accents (e.g., 'é' -> 'e') and remove non-ASCII characters
    text = unicodedata.normalize('NFKD', text).encode('ascii', 'ignore').decode('utf-8')

    # Remove Special Characters: Keep only letters, numbers, and spaces
    text = re.sub(r'[^a-zA-Z0-9\s]', '', text)

    # Clean up extra whitespaces left behind
    text = re.sub(r'\s+', ' ', text).strip()

    return text
```

```python
df['Sentence'] = df['Sentence'].apply(clean_text)
```

```python
df['Sentence'].count()
```

```python
df.head()
```

```python
sentiment_counts = df.groupby('Sentiment').size()
print(sentiment_counts)
```

```python
from sklearn.model_selection import train_test_split

X = df['Sentence']
y = df['Sentiment']

# Stratified split: 80% train, 20% test
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    stratify=y,
    random_state=42
)
```

```python
print(y_train.value_counts())
```

```python
print(y_test.value_counts())
```

```python
import pandas as pd

train_df = pd.DataFrame({'Sentence': X_train, 'Sentiment': y_train})
test_df = pd.DataFrame({'Sentence': X_test, 'Sentiment': y_test})

train_df.to_parquet('train_data.parquet', index=False)
test_df.to_parquet('test_data.parquet', index=False)
```
